```python
lead_ii = signals[:,1]

plt.figure(figsize=(15,4))
plt.plot(lead_ii[:5000])
plt.title("Lead II")
plt.show()
```

```python
fig, axes = plt.subplots(12,1, figsize=(15,20))

for i in range(12):
    
    axes[i].plot(signals[:5000, i])
    axes[i].set_title(record.sig_name[i])

plt.tight_layout()
plt.show()
```

```python
print(record.comments)
```

```python
def get_mi_location(comments):

    text = " ".join(comments).lower()

    if "anterior" in text:
        return "anterior"

    elif "inferior" in text:
        return "inferior"

    elif "lateral" in text:
        return "lateral"

    elif "posterior" in text:
        return "posterior"

    else:
        return None
```

```python
label = get_mi_location(record.comments)

print(label)
```
